# Task 4 
# Data Visualization

## 5 VIsualiztion For Each Dataset

```python
import seaborn as sns 
import matplotlib.pyplot as plt
%matplotlib inline
```

# Bar Chart

```python

top_countries=df["country"].value_counts().head(5).index
df_top5=df[df["country"].isin(top_countries)]
country_type=pd.crosstab(df_top5["country"],df_top5["type"])
country_type.plot(kind="bar",figsize=(10,6),color=["darkred","black"],edgecolor='black')
plt.title("Netflix Content Type Distribution in Top 5 Countries",fontsize=14,fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Count Of Movies/ Tv Shows")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Findings:

#### US & India: High movie-dominant markets,with movies significantly outnumbering TV shows.
#### Japan & South Korea: TV show-dominant markets,driven by the global popularity of anime and k-Darmas.
#### United Kingdom: A balanced market with an almost equal split between movies and TV shows.

# Histogram

```python
plt.figure(figsize=(10,6))
sns.histplot(df["release_year"],bins=30,kde=True, color="darkred")
plt.title("Distribution of Content Release years on Netflix",fontsize=14, fontweight="bold")
plt.xlabel("Release Year")
plt.ylabel("Count of Movies / TV Shows")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

# Findings

### Recent Surge: The dataset is heavily left-skewed showing that the vast majority of netflix content was released after 2010.
### Peak Era:  Content production peaked dramatically between 2015 and 2020, Netflix aggressive expansion into original programming.
### Historical Content: There is very minimal content from the 1940s to the 1980s indicating that Netflix focuses primarily on modern media.

# Pie Chart

```python
plt.figure(figsize=(6,6))
content_counts=df["type"].value_counts()
wedges, texts, autotexts =plt.pie(content_counts,
        labels=content_counts.index,
       autopct="%1.1f%%",
startangle=90,
colors=["darkred","black"],
       pctdistance=0.6,
       labeldistance=1.1)
for text in texts:
    text.set_color("black")
    text.set_fontsize(12)
    text.set_weight("bold")
for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontsize(12)
    autotext.set_weight("bold")
plt.title("Netflix Content distribution (Movies vs TV Shows)", fontsize=14, fontweight="bold",pad=20)
plt.show()          
```

 # Findings

### Dominant Content Type: Movies make up the majority of Netflix's platform at 69.6%. while TV shows account for the remaining 30.4%.
### Bussiness Insight: This indicates that Netflix's business strategy is heavily focused on full-length feature films, although TV shows still maintain a significant nearly one-third share of the library.

# Line Chart

```python
df_modern=df[df["release_year"]>=2000]
yearly_trend=df_modern.groupby("release_year").size()

plt.figure(figsize=(10,5))

plt.plot(yearly_trend.index, yearly_trend.values, marker="o", color="darkred", linewidth=2.5)

plt.title("Netflix Content Release Trend (2000 2021)",fontsize=14 ,fontweight="bold")
plt.xlabel("Release Year")
plt.ylabel("Total Movies / TV Shows Released")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Finding

### Growth Trend: The line chart clearly illustrates exponential growth in Netflix content release starting around 2015, peaking dramatically near 2018-2019
### Drop Insights: A slight drop after 2019/2020 might reflect production slowdowns due to global events (Like COVID-19 Lockdowns).

# Box Plot

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="type", y= "release_year", data=df ,hue="type", palette=["darkred","gray"],legend=False)
plt.title("Distribution of Release Years by Content Type (Outlier Analysis)", fontsize=14, fontweight="bold")
plt.xlabel("Content Type")
plt.ylabel("Release Year")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

# Finding
## Statistical  Insisghts: Release Year Distribution & Outliers
### Data Concentration: The compact boxes located at the top (between 2015 and 2021) indicate that more than 60% of Netflix's entire catalog consists of very recent releases.
### Presence of Outliers: The long tail of individual dots stretching downwards shows a heavy presence of outliers for both Movies and Tv shows. This proves that Netflix maintain a significant collection of classis media dating as far back as the 1940s.
### TV Shows vs Movies: The TV Show is slightly more concept at the top compared to Movies, meaning Netflix's TV Show library is even more modern and newly produced than its Movie library
