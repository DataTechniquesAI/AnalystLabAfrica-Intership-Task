### Distribution of important Variables

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df["age"],kde=True,bins=30)
plt.title("Age Distribution",fontweight="bold")
plt.xlabel("Age")
plt.ylabel("frequency")
plt.show()
```

```python
sns.boxplot(y=df["balance"],color="orange")
plt.title("Boxplot of balance () Cheecking Outlliers",fontweight="bold")
plt.ylabel("Balance")
plt.show()
```

```python
print(df["job"].value_counts())
```

```python
print(df["job"].value_counts(normalize=True)*100)
```

```python
sns.countplot(data=df, x = "marital", palette="viridis",legend=False)
plt.title("Distribution of Material Status",fontweight="bold")
plt.xlabel("Material Status")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

### Trend and Patterns

```python
corr=df.select_dtypes(include=["float64","int64"]).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap",fontweight="bold")
plt.show()
```

### Categorical Trends

```python
sns.countplot(data =df,x="job", hue= "deposit")
plt.xticks(rotation=45)
plt.title("Compaign Success by Job Type",fontweight="bold")
plt.xlabel("job Type")
plt.ylabel("Count")
plt.show()
```

# Bar Chart

```python
plt.figure(figsize=(8,5))
sns.countplot(data= df,x="marital",hue="marital",palette="viridis" , legend =False)
plt.title("Distribution of Marital Status",fontweight= "bold")
plt.xlabel("Marital status")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

# Histogram

```python
plt.figure(figsize=(10,8))
sns.histplot(df["age"],kde=True,bins=30,color="blue")
plt.title("Histogram Distribution of Age", fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

# Box PLot

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["balance"],color="orange")
plt.title("Customer Balance",fontweight="bold")
plt.ylabel("balance")
plt.show()
```

# Pie Chart

```python
deposit_counts=df["deposit"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(
    deposit_counts,
    labels=deposit_counts.index,
    autopct="%1.f%%",
    startangle=90,
    colors=["skyblue","lightcoral"],
)
plt.title("Deposit Subscriptions Proportion ",fontweight="bold")
plt.show()
```

# Line Charts

```python
day_counts=df["day"].value_counts().sort_index()
plt.figure(figsize=(10,8))
plt.plot(day_counts.index,day_counts.values, marker='o',color="green",linestyle="-")
plt.title("Campaign Activity Trend By Days",fontweight="bold")
plt.xlabel("Day of the Month")
plt.ylabel("Number Of contact")
plt.grid(True)
plt.show()
```

# Correlation HeatMap

```python
plt.figure(figsize=(10,8))
corr=df.select_dtypes(include=["float64","int64"]).corr()
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation",fontweight="bold")
plt.show()
```

# Identify Key drivers

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df,x='job',hue='deposit',palette="Set2")
plt.title("Deposit Sucess By Job Type",fontweight="bold")
plt.xlabel("Job Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

# Education Level

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,8))
sns.countplot(data=df, x="education" , hue="deposit", palette="Set1")
plt.title("Deposit Sucess ny Education Level",fontweight="bold")
plt.xlabel("Education Level")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

# Contact Method

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df , x="contact",hue="deposit",palette="viridis")
plt.title("Deposit Success by Contact Method ")
plt.xlabel("Contact Method")
plt.ylabel("Count")
plt.show()
```
