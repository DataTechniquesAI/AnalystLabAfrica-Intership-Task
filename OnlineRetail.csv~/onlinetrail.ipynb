## Data Visualization

#### 5 Visualization For Each Dataset

```python
import seaborn as sns 
import matplotlib.pyplot as plt
%matplotlib inline
```

# Bar Chart 
### Top 20 Countries

```python
plt.figure(figsize=(10,5))
top_countries= df.groupby("Country")["Revenue"].sum().head(20)

sns.barplot(x=top_countries.index, y=top_countries.values,hue=top_countries.index, palette="viridis",legend=False)
#Title
plt.title("TOP 20 Countries by Total Revenue",fontsize=14,fontweight="bold")
plt.xlabel("Country",fontsize=12)
plt.ylabel("Total Revenue",fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

# Findings:
#### Dominant Market:   The bar chart clearly indicates that the United Kingdom (UK) is the leading market,generating exponentially higher total revenue compare to all other nations.
#### Secondary Contribution:   Following the UK,Countries like Germany,France, and the NetherLands emerge as the next tier of revenue contributors, while the remaining countries show relatively minimal total revenue impact

# Histogram

```python
plt.figure(figsize=(10,5))
filtered_price=df[(df["UnitPrice"]>0) & (df["UnitPrice"]<=10)]
sns.histplot(filtered_price["UnitPrice"],bins=30,kde=True,color="purple")
plt.title("Distribution of Product Unit Prices(Filtered<=10)",fontsize=14,fontweight="bold")
plt.xlabel("Unit Price",fontsize=12)
plt.ylabel("Country of Product",fontsize=12)
plt.show()
```

# Findings
### Price Concentration: The distribution show a heavy right-skew, revealiing that the vast majority of products are low -cost items concentrated between 0 and 3. the highest peak is observed around the 1 and 2 price range.
### Consumer behaviour:  This suggests that the e-commerce platform's inventory and sales volume are driven primarily by affordable , low-unit-price commodities, with product counts dropping drastically as unit prices exceed 6.

# Pie Chart

```python
plt.figure(figsize=(10,8))
top_countries=df.groupby("Country")["Revenue"].sum().sort_values(ascending=False).head(5)
plt.pie(top_countries.values,
       labels=top_countries.index,
       autopct="1%.1f%%",
        startangle=140,
        colors=sns.color_palette("viridis",len(top_countries)))
plt.title("Revenue Share of Top 5 Countries",fontsize=14,fontweight="bold")

plt.show()      
```

# Findings:
### Massive Market Monopolization: Among the top 5 countries , the United kingdom uniquely commands a staggering **89.4%** of the totla revenue share, highlighting the business's extreme reliance on the UK market.
### Minor Stakeholders: The remaining top countries-the Netherlands(2.9%), eire (2.5%), Germany, and France-collectively account for just over 10% of the share, emphasizing their minor role in comparison to the primary Uk market.

# Line Chart

```python
plt.figure(figsize=(6,6))
top_countries=df.groupby("Country")["Revenue"].sum().head(10)
plt.plot(top_countries.index,top_countries.values)
plt.title("Top Countries",fontweight="bold")
plt.xticks(rotation=45)
plt.show()
```

# Findings:
### Revenue fluctuations: This line chart tracks and highlights the sharp peaks and valleys in revenue genneration across different countries.
### Notable Spikes: Visible upward spikes can be seen for specific regions like Australia and Belgium , whereas the flat trendline for most other countries indicates a consistently low per-country purchase volume across the board

# Box Plot

```python
plt.figure(figsize=(10,12))
sns.boxplot(x=df["Revenue"],y=df["Country"],showfliers=False)
plt.xlabel("Revenue")
plt.ylabel("Country")
plt.title("Revenue Boxplot",fontweight="bold",fontsize=14)
plt.show()
```

# Findings:
### Spread and Variability: Countries such as the Netherlands , Australia, and japan display wider boxes and extended whiskers. This indicates a high variance and spread in transaction values, meaning custommers in these regions frequently place higher-value orders.
### Stable, Lower,Value: Conversely, the United Kingdom and the majority of other nations show highly compressed boxes close to zero, demonstrating a steady pattern of consistent, lower-value transactions.
